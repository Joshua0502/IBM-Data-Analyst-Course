### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
print('Number of unique countries:', df['Country'].nunique())
print('\nTop 10 Countries:\n', df['Country'].value_counts().head(10))
print('\nBottom 10 Countries:\n', df['Country'].value_counts().tail(10))
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
country_replacement = {
    'United States of America' : 'USA',
    'United Kingdom of Great Britain and Northern Ireland' : 'UK',
    'Iran, Islamic Republic of...' : 'Iran',
    'Venezuela, Bolivarian Republic of...' : 'Venezuela',
    'Viet Nam' : 'Vietnam',
    'Republic of Korea' : 'South Korea',
    'Democratic People\'s Republic of Korea' : 'North Korea',
    'Congo, Republic of the...' : 'Republic of Congo',
    'Nomadic' : 'np.nan'
}

df['Country'] = df['Country'].replace(country_replacement)
df['Country'].unique()
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
df['Employment'].unique()
```

```python
## Write your code here
# encoded_employment = pd.get_dummies(df['Employment'], prefix = 'Employment')

# df = pd.concat([df, encoded_employment], axis=1)
# df.head(10)

employment_split_encoded = df['Employment'].str.get_dummies(sep=';')

# 2. Clean up the resulting column names (strip whitespace) and add a prefix for clarity.
employment_split_encoded.columns = ['Employment_' + col.strip() for col in employment_split_encoded.columns]

# 3. Concatenate the new encoded columns back to the main DataFrame
df = pd.concat([df, employment_split_encoded], axis=1)

# 4. Drop the original 'Employment' column
#df.drop('Employment', axis=1, inplace=True)

df.head(10)
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
df.isna().sum().idxmax()
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
df['ConvertedCompYearly'].isna().sum()
```

```python
## Write your code here
comp_mean = df['ConvertedCompYearly'].mean()

df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(comp_mean)

df['ConvertedCompYearly'].isna().sum()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
remote_mode = df['RemoteWork'].mode()[0]

df['RemoteWork'] = df['RemoteWork'].fillna(remote_mode)

df['RemoteWork'].isna().sum()
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
comp_min = df['ConvertedCompYearly'].min()
comp_max = df['ConvertedCompYearly'].max()

df['ConvertedCompYearly_MinMax'] = ((df['ConvertedCompYearly'] - comp_min) / (comp_max - comp_min))

df['ConvertedCompYearly_MinMax'].head(20)
```

```python
print("Max Value: ", comp_max)
print("Min Value: ", comp_min)
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

# Natural log
#df['ConvertedCompYearly_LofTransform'] = np.log1p(df['ConvertedCompYearly'])

# log of 1 plus the value: for better handing 0 values
df['ConvertedCompYearly_LogTransform'] = np.log1p(df['ConvertedCompYearly'])

df['ConvertedCompYearly_LogTransform'].head()
```

```python
df[['ConvertedCompYearly', 'ConvertedCompYearly_MinMax', 'ConvertedCompYearly_LogTransform']].head(10)
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
df['YearsCodePro'].value_counts()
```

```python
## Write your code here

# Junior: Less than 3 years of professional experience.
# Mid-Level: 3 to 10 years of professional experience.
# Senior/Expert: More than 10 years of professional experience.

df['YearsCodePro'].replace({'Less than 1 year' : 0.5, 'More than 50 years' : 51}, inplace = True)

df['YearsCodePro_Cleaned'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')`

bins = [0, 3, 10, df['YearsCodePro_Cleaned'].max() + 1]
labels = ['Junior', 'Mid-Level', 'Senior']

df['ExperienceLevel'] = pd.cut(
    df['YearsCodePro_Cleaned'], bins = bins, label = labels, right = False
)

df['ExperienceLevel'].value_counts()
```

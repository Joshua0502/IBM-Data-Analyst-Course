## Identify and Analyze Duplicates


### Task 1: Identify Duplicate Rows

```python
## Write your code here
print("There are", df.duplicated().sum(), "duplicated rows")

duplicates = df[df.duplicated()]
duplicates.head()
```

### Task 2: Analyze Characteristics of Duplicates
1. Identify duplicate rows based on selected columns such as MainBranch, Employment, and RemoteWork. Analyse which columns frequently contain identical values within these duplicate rows.
2. Analyse the characteristics of rows that are duplicates based on a subset of columns, such as MainBranch, Employment, and RemoteWork. Determine which columns frequently have identical values across these rows.
   


```python
## Write your code here
print('Duplicates:', df[['MainBranch', 'Employment', 'RemoteWork']].duplicated().sum())
selected_duplicates = df[df[['MainBranch', 'Employment', 'RemoteWork']].duplicated()]
selected_duplicates.head()
```

```python
analysis_cols = [col for col in df.columns if col not in ['MainBranch', 'Employment', 'RemoteWork']]

print("\n--- Uniqueness Analysis of Other Columns ---")
for col in analysis_cols:
    num_unique = selected_duplicates[col].nunique()
    total_rows = len(selected_duplicates)
    
    # Analyze the ratio of unique values to the total number of duplicate rows
    print(f"Column '{col}': {num_unique} unique values out of {total_rows} rows.")
    # A low ratio means high identicality.
```

### Task 3: Visualize Duplicates Distribution
1. Create visualizations to show the distribution of duplicates across different categories.
2. Use bar charts or pie charts to represent the distribution of duplicates by Country and Employment.


```python
## Write your code here
# 1. Get the distribution data
country_distribution = df['Country'].value_counts()

# 2. Select the Top N (e.g., Top 15)
top_n = 15
top_countries = country_distribution.head(top_n)

# 3. Create the focused bar chart
plt.figure(figsize=(12, 7)) # Increase figure size for better spacing
top_countries.plot(kind='bar', color='skyblue')
plt.title(f'Distribution of Subset Duplicates by Country (Top {top_n})')
plt.xlabel('Country')
plt.ylabel('Count of Duplicates')
plt.xticks(rotation=45, ha='right') # Rotate labels for better fit
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Automatically adjusts subplot params for a tight layout
plt.show()
```

```python
employment_distribution = df['Employment'].value_counts()

#Create the Pie Chart
plt.figure(figsize=(8, 8))
plt.pie(
    # Values (The sizes of the slices)
    employment_distribution.values, 
    # Labels (The actual employment types, which are in the index)
    labels=employment_distribution.index, 
    # Automatically display the percentage for each slice
    autopct='%1.1f%%', 
    # Optional: Start the first slice at the top
    startangle=90 
)
plt.title('Distribution of Employment Types')
plt.axis('equal') # Ensures the pie chart is drawn as a circle
plt.show()
```

### Task 4: Strategic Removal of Duplicates
1. Decide which columns are critical for defining uniqueness in the dataset.
2. Remove duplicates based on a subset of columns if complete row duplication is not a good criterion.


```python
## Write your code here
# --- 1. Define the Critical Uniqueness Columns ---
# These columns must be identical for a row to be considered a duplicate entry.
uniqueness_subset = [
    'MainBranch', 
    'Employment', 
    'RemoteWork', 
    'Country', 
    'DevType', 
    'YearsCode'
]

# --- 2. Remove Duplicates Based on the Subset ---

# Count the number of duplicates BEFORE removal for reporting
duplicates_to_remove = df.duplicated(subset=uniqueness_subset, keep='first').sum()

# Create the new DataFrame by dropping the duplicates
# We use keep='first' to retain the first (original) entry and drop the subsequent copies.
df_clean = df.drop_duplicates(
    subset=uniqueness_subset, 
    keep='first',
    inplace=False # ensure a new DataFrame is returned
)

# --- 3. Report the Results ---

print(f"Original DataFrame Shape: {df.shape}")
print(f"Number of duplicates found (based on the strategic subset): {duplicates_to_remove}")
print(f"Cleaned DataFrame Shape: {df_clean.shape}")
print("\nDuplicate removal complete. The cleaned data is stored in 'df_clean'.")
```

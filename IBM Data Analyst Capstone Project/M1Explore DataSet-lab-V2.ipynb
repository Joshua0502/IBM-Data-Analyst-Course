Print the mean age of the survey participants.


```python
## Write your code here

# 1. Create a mapping dictionary for age range midpoints
age_midpoints = {
    'Under 18 years old': 8.5,    # (0 + 17) / 2
    '18-24 years old': 21.0,       # (18 + 24) / 2
    '25-34 years old': 29.5,
    '35-44 years old': 39.5,       # (35 + 44) / 2
    '45-54 years old': 49.5,     # (45 + 54) / 2
    '55-64 years old': 59.5,
    'Prefer not to say'
    # You would need to add any other age ranges present in your full dataset
}

# 2. Map the age ranges to their numerical midpoints
df['Age_Midpoint'] = df['Age'].map(age_midpoints)

# 3. Calculate the estimated mean of the new midpoint column
estimated_mean_age = df['Age_Midpoint'].mean()

print(df)
print(f"\nThe Estimated Mean Age is: {estimated_mean_age:.2f} years")

print("Mean Age: ",df['Age'].mean().round(1))
```

The dataset is the result of a world wide survey. Print how many unique countries are there in the Country column.


```python
## Write your code here
unique_countries = df['Country'].unique().tolist()
print("There are", df['Country'].nunique(), "unique countries:")
count = 1
for i in unique_countries:
    print(count,"- ", i)
    count = count + 1
```

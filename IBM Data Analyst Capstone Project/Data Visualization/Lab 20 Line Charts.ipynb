### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
## Write your code here

# 1. Prepare the Data
# Ensure ConvertedCompYearly is numeric
df['ConvertedCompYearly'] = pd.to_numeric(df['ConvertedCompYearly'], errors='coerce')

# Define the logical chronological order for Age groups
age_order = [
    'Under 18 years old', 
    '18-24 years old', 
    '25-34 years old', 
    '35-44 years old', 
    '45-54 years old', 
    '55-64 years old', 
    '65 years or older'
]

# 2. Aggregation
# Group by Age and calculate the median compensation
# We reindex to ensure the age groups follow the logical life-cycle order
age_comp_trends = df.groupby('Age')['ConvertedCompYearly'].median().reindex(age_order)

# Remove categories with no data if necessary
age_comp_trends = age_comp_trends.dropna()

# 3. Create the Line Chart
plt.figure(figsize=(12, 6))

plt.plot(
    age_comp_trends.index, 
    age_comp_trends.values, 
    marker='o',           # Circle markers at data points
    linestyle='-',        # Solid line
    linewidth=2, 
    color='#2ca02c',      # Professional green color
    markersize=8,
    markerfacecolor='white',
    markeredgewidth=2
)

# 4. Formatting and Styling
plt.title('Trend of Median Yearly Compensation by Age Group', fontsize=16, pad=20)
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Median Annual Compensation (USD)', fontsize=12)

# Prevent scientific notation on Y-axis
plt.ticklabel_format(style='plain', axis='y')

# Add gridlines for easier value estimation
plt.grid(True, linestyle='--', alpha=0.6)

# Rotate X-axis labels for better readability
plt.xticks(rotation=30, ha='right')

plt.tight_layout()
plt.show()

# Print summary for reference
print("Median Compensation by Age Group:")
print(age_comp_trends.apply(lambda x: f"${x:,.2f}"))
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
## Write your code here

# 1. Prepare the Data
# Ensure Job Satisfaction (JobSatPoints_6) and Experience are numeric
df['JobSatPoints_6'] = pd.to_numeric(df['JobSatPoints_6'], errors='coerce')

# Clean 'YearsCodePro' column to get numeric experience
yearscodepro_map = {
    'Less than 1 year': 0.5,
    'More than 50 years': 52.0
}
df['YearsCodePro_Numeric'] = df['YearsCodePro'].replace(yearscodepro_map)
df['YearsCodePro_Numeric'] = pd.to_numeric(df['YearsCodePro_Numeric'], errors='coerce')

# 2. Define Experience Bins
# Grouping experience into logical tiers (0-5, 5-10, 10-20, and 20+ years)
bins = [0, 5, 10, 20, 100]
labels = ['0-5 Years', '5-10 Years', '10-20 Years', '>20 Years']
df['Experience_Level'] = pd.cut(
    df['YearsCodePro_Numeric'], 
    bins=bins, 
    labels=labels, 
    right=False
)

# 3. Aggregation
# Group by the new Experience Level category and calculate median satisfaction
# The categorical type ensures the x-axis follows the logical order
exp_satisfaction_trends = df.groupby('Experience_Level')['JobSatPoints_6'].median()

# Remove categories with no data if necessary
exp_satisfaction_trends = exp_satisfaction_trends.dropna()

# 4. Create the Line Chart
plt.figure(figsize=(10, 6))

plt.plot(
    exp_satisfaction_trends.index.astype(str), 
    exp_satisfaction_trends.values, 
    marker='D',           # Diamond markers at data points
    linestyle='-',        # Solid line
    linewidth=3, 
    color='#e67e22',      # Orange color to represent satisfaction
    markersize=10,
    markerfacecolor='white',
    markeredgewidth=2
)

# 5. Formatting and Styling
plt.title('Median Job Satisfaction Trend by Experience Level', fontsize=16, pad=20)
plt.xlabel('Professional Coding Experience', fontsize=12)
plt.ylabel('Median Job Satisfaction (JobSatPoints_6)', fontsize=12)

# Set Y-axis limits based on expected satisfaction range (e.g., 0-10)
plt.ylim(0, 10.5)

# Add gridlines for easier value estimation
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()

# Print summary for reference
print("Median Job Satisfaction by Experience Level:")
print(exp_satisfaction_trends)
```

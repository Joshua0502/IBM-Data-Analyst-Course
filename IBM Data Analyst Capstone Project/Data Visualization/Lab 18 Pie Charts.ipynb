### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
# 1. Prepare the data
# Drop rows with missing values in the DatabaseWantToWorkWith column
db_data = df['DatabaseWantToWorkWith'].dropna()

# 2. Process multi-select values
# Split the semicolon-separated strings and explode into a single series of database names
all_databases = db_data.str.split(';').explode()

# 3. Get the top 5 most frequent databases
top_5_databases = all_databases.value_counts().head(5)

# 4. Create the Pie Chart
plt.figure(figsize=(10, 8))

# Define a nice color palette
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0']

plt.pie(
    top_5_databases, 
    labels=top_5_databases.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=colors,
    pctdistance=0.85,
    explode=(0.05, 0, 0, 0, 0) # Slightly explode the top one
);

# 5. Add a hole in the center to make it a Donut Chart (optional, for aesthetics)
centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Top 5 Databases Respondents Want to Work With', fontsize=15)
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()

# Print the raw counts for reference
print("Top 5 Desired Databases (Counts):")
print(top_5_databases)
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.

### Task 3: Analyzing and Interpreting Composition


In this task, you will create additional pie charts to analyze specific aspects of the survey data. Use `pandas` and `matplotlib` to complete each task and interpret the findings.



##### 3.1 Pie Chart of `Respondents` Most Admired Programming Languages


The `LanguageAdmired` column lists the programming languages respondents admire most. Create a pie chart to visualize the top 5 admired languages.



```python
##Write your code here

LanguageAdmired = df['LanguageAdmired'].str.split(';').explode()
LanguageAdmired = LanguageAdmired.dropna()

top5_LanguageAdmired = LanguageAdmired.value_counts().head(5)

plt.figure(figsize=(10,8))

plt.pie(
    top5_LanguageAdmired,
    labels = top5_LanguageAdmired.index,
    autopct = '%1.1f%%'
);

plt.title('Top 5 most admired programming languages')
plt.tight_layout()
plt.show()
```

```python
# 1. Prepare the data
# Drop rows with missing values in the LanguageAdmired column
lang_data = df['LanguageAdmired'].dropna()

# 2. Process multi-select values
# Split the semicolon-separated strings (e.g., "Rust;TypeScript;Python") 
# and explode into a single series of language names
all_languages = lang_data.str.split(';').explode()

# 3. Get the top 5 most frequent languages
top_5_admired = all_languages.value_counts().head(5)

# 4. Create the Pie Chart
plt.figure(figsize=(10, 8))

# Define a professional color palette
# Using a slightly different palette to distinguish from the database analysis
colors = ['#4e79a7', '#f28e2c', '#e15759', '#76b7b2', '#59a14f']

plt.pie(
    top_5_admired, 
    labels=top_5_admired.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=colors,
    pctdistance=0.85,
    explode=(0.1, 0, 0, 0, 0) # Highlight the #1 most admired language
)

# 5. Add a hole in the center for a Donut Chart effect
centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Top 5 Most Admired Programming Languages', fontsize=16, pad=20)
plt.axis('equal') # Ensures the pie chart is a perfect circle
plt.tight_layout()
plt.show()

# 6. Interpretation of Findings
print("--- Analysis of Admired Languages ---")
print(f"The top language is '{top_5_admired.index[0]}' with {top_5_admired.iloc[0]} admirers.")
print("\nTop 5 Breakdown:")
print(top_5_admired)
```
